```python
from selenium import webdriver
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome()
driver.get("https://www.ceazamet.cl/pronostico/")

time.sleep(5)

html = driver.page_source
sopa = BeautifulSoup(html, "html.parser")

resultados = sopa.select("span.temperaturas")

for r in resultados:
    print(r.text.strip())

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# Configurar opciones de Chrome
options = Options()
options.add_argument("--headless")        # Ejecutar sin abrir navegador
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(options=options)

driver.get("https://www.ceazamet.cl/pronostico/")
time.sleep(5)

html = driver.page_source
sopa = BeautifulSoup(html, "html.parser")

resultados = sopa.select("span.temperaturas")

for r in resultados:
    print(r.text.strip())

driver.quit()
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.ceazamet.cl/pronostico/"
headers = {"User-Agent": "Mozilla/5.0"}

respuesta = requests.get(url, headers=headers)

sopa = BeautifulSoup(respuesta.text, "html.parser")

# buscar temperaturas
temps = sopa.select(".temperaturas")

for t in temps:
    print(t.text.strip())
```

```python
import requests 
from bs4 import BeautifulSoup 

# Configuraci n de entrada 
url = "https://www.eciem.cl/anuncios/pregrados/1/todos" 
headers = {"User-Agent": "Mozilla/5.0"} # Para que no nos bloqueen 

# 1. Petici n al servidor 
respuesta = requests.get(url, headers=headers) 

# 2. Convertir el HTML en un objeto de Python 
sopa = BeautifulSoup(respuesta.text, 'html.parser') 

# 3. Buscar los datos (Ejemplo: buscar todos los t tulos) 
# Cambia ’h2’ y ’nombre-de-la-clase’ por lo que inspeccionaste 
resultados = sopa.find_all('span', class_='fechas') 

# 4. Mostrar en pantalla 
for dato in resultados: 
    print(f"Encontrado: {dato.text.strip()}")
```
